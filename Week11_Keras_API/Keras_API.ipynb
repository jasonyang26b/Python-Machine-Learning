## Keras Functional API & Keras Sequential API

The Keras Functional API is a flexible way to build neural network architectures in TensorFlow/Keras. Unlike the Sequential API, which only allows for a linear stack of layers, the Functional API enables you to create complex network topologies such as:

*   **Multiple inputs:** Networks that accept data from several different sources.
*   **Multiple outputs:** Networks designed for multi-task learning (e.g., classification and regression simultaneously).
*   **Shared layers:** Reusing the same layer's output in multiple places within the model.
*   **Non-linear connections:** Creating branching paths and skip connections (like in ResNets).

### Key Concepts

1.  **Inputs:** You must explicitly define the input tensors using `tf.keras.Input()`.
2.  **Layers:** Layers are called on the input tensor, and they return a new output tensor.
3.  **Model Construction:** You define the graph by connecting these input and output tensors to your final `tf.keras.Model` instance using the functional signature: `model = tf.keras.Model(inputs=input_tensor, outputs=output_tensor)`.

### Basic Example: A Simple Network

This example shows a basic feed-forward network using the Functional API.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# 1. Define the input layer (specifying the shape of the data)
input_tensor = keras.Input(shape=(784,))

# 2. Define the layers and connect them sequentially
x = layers.Dense(128, activation='relu')(input_tensor)
x = layers.Dropout(0.3)(x)
output_tensor = layers.Dense(10, activation='softmax')(x)

# 3. Create the Model instance
model = keras.Model(inputs=input_tensor, outputs=output_tensor)

# Compile and summarize
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

### Advanced Example: Multiple Inputs and Merging

This example demonstrates a model that takes two different types of input (e.g., image features and metadata) and combines them before the final prediction.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# --- Define Inputs ---
# Input 1: Image data (e.g., MNIST-like)
input_image = keras.Input(shape=(28, 28, 1), name='image_input')

# Input 2: Metadata (e.g., age, gender)
input_metadata = keras.Input(shape=(5,), name='metadata_input')

# --- Process Image Branch ---
x_image = layers.Flatten()(input_image)
x_image = layers.Dense(64, activation='relu')(x_image)
x_image = layers.Dropout(0.2)(x_image)

# --- Process Metadata Branch ---
x_metadata = layers.Dense(32, activation='relu')(input_metadata)

# --- Merge Branches ---
# Concatenate the processed features from both branches
merged = layers.concatenate([x_image, x_metadata])

# --- Final Layers ---
z = layers.Dense(32, activation='relu')(merged)
output_tensor = layers.Dense(1, activation='sigmoid', name='output')(z)

# --- Create the Model ---
model_multi = keras.Model(inputs=[input_image, input_metadata], outputs=output_tensor)

model_multi.compile(optimizer='adam',
                    loss='binary_crossentropy',
                    metrics=['accuracy'])

model_multi.summary()
```

### When to Use Functional API vs. Sequential API

| Feature | `Sequential` API | `Functional` API |
| :--- | :--- | :--- |
| **Structure** | Linear stack of layers. | Graph-based; allows complex topologies. |
| **Inputs/Outputs** | Single input, single output. | Multiple inputs, multiple outputs supported. |
| **Complexity** | Simple models (e.g., basic CNNs, MLPs). | Complex architectures (ResNets, Siamese Networks, multi-input models). |
| **Usage** | Quick prototyping for simple tasks. | Production-level, complex research models. |
