## **Hierarchical Clustering:**

Hierarchical clustering is a method of organizing data into a tree-like structure (called a dendrogram) by grouping similar items together at different levels of granularity.

## How it works

There are two main approaches:

**Agglomerative (bottom-up)**: Start with each data point as its own cluster, then repeatedly merge the two closest clusters until everything is in one cluster.

**Divisive (top-down)**: Start with all data in one cluster, then repeatedly split clusters into smaller ones until each point is isolated.

## Key concepts

**Distance metric**: Determines how "close" two items are (Euclidean, Manhattan, correlation, etc.)

**Linkage criterion**: Defines how to measure distance between clusters:
- *Single linkage*: Distance between closest points in two clusters
- *Complete linkage*: Distance between farthest points
- *Average linkage*: Average distance between all pairs
- *Ward's*: Minimizes variance within clusters (often preferred)

## Dendrogram

The output is a tree diagram showing the hierarchical structure. You can cut the tree at different heights to get different numbers of clusters — higher cuts = fewer, larger clusters; lower cuts = more, smaller clusters.

## Advantages & disadvantages

**Pros:**
- No need to specify cluster count upfront
- Interpretable tree structure
- Works with any distance metric (Minkowski; Euclidean; Manhatan; Jaccard; Hamming; Cosine; Mahalanobis)

**Cons:**
- Computationally expensive (O(n²) or worse)
- Decisions are irreversible (can't undo merges)
- Sensitive to outliers and noise
- Slower than k-means for large datasets

## Use cases

Good for **exploratory analysis**, **taxonomy building**, **gene sequence analysis**, and when you want to understand relationships at multiple scales.

## **Why Hierarchical Clustering computationally expensive:**

Hierarchical clustering, specifically the agglomerative approach, is computationally expensive because it requires calculating the proximity (distance) between every pair of points in the dataset to determine which two should be merged next.

The complexity arises from the following factors:

1.  **Distance Matrix Calculation:** To begin, the algorithm must compute a distance matrix containing the distances between all $n$ points. Since there are $\frac{n(n-1)}{2}$ unique pairs, this initial step is $O(n^2)$.
2.  **Iterative Merging:** The algorithm performs $n-1$ iterations. In each iteration, it must search through the existing distances to find the minimum value (the closest pair).
3.  **Updating the Matrix:** Every time two clusters are merged into one, the distances between the new cluster and all other remaining clusters must be recalculated or updated. This update step must be performed for the remaining clusters in each of the $n$ iterations.

While optimized algorithms (like using priority queues or specialized data structures) can reduce the complexity for certain linkage methods, the fundamental requirement to compare elements against one another ensures that the complexity remains at least $O(n^2)$ for most implementations.
