```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_samples, silhouette_score

from scipy.spatial.distance import pdist, squareform
```

#### **pdist** and **squareform**

```python
X, y = make_blobs(n_samples=150, n_features=2, centers=3, 
           cluster_std=2.5, shuffle=True, random_state=42)
```

```python
X.shape
```

```python
# calculate the pairwise distance
distance = pdist(X, metric='euclidean')
len(distance)
```

```python
# confirm total pairwise distance
(150 - 1) * 150 / 2
```

```python
# convert from list to square form
square_distance = squareform(distance)
square_distance.shape
```

#### How to use **set_link_color_palette**:

```python
from scipy.cluster.hierarchy import set_link_color_palette, dendrogram, linkage
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
X = np.random.randn(30, 2)
Z = linkage(X, method='ward')

# Use named matplotlib colors
set_link_color_palette(['red', 'blue', 'green', 'purple', 'orange'])

plt.figure(figsize=(12, 6))
dendrogram(Z, color_threshold=2)
plt.title('Dendrogram with Named Colors')
plt.show()
```
