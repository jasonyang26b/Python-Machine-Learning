## **Minkowski distance:**

Minkowski distance is a generalized distance metric defined by the formula:

$$D(x, y) = \left( \sum_{i=1}^{n} |x_i - y_i|^p \right)^{1/p}$$

Where:
*   **$x$ and $y$** are two points in $n$-dimensional space.
*   **$p$** is a positive real number that determines the shape of the distance measure.

### Common Special Cases
The value of $p$ determines which specific distance metric is being used:

*   **$p = 1$ (Manhattan Distance / L1 Norm):** Calculates the sum of the absolute differences of their coordinates. It represents distance traveled on a grid (city block distance).
*   **$p = 2$ (Euclidean Distance / L2 Norm):** The standard "straight-line" distance between two points in Euclidean space.
*   **$p \to \infty$ (Chebyshev Distance / L∞ Norm):** The distance is defined as the maximum absolute difference between any single dimension: $D(x, y) = \max(|x_i - y_i|)$.

### Characteristics in Hierarchical Clustering
*   **Parameter Sensitivity:** The choice of $p$ significantly alters the resulting cluster structure. Lower values of $p$ (like $p=1$) are more robust to outliers, while higher values (like $p=2$) are more sensitive to large differences in a single dimension.
*   **Computational Complexity:** Calculating Minkowski distance for large datasets can be computationally expensive, especially as $p$ increases, due to the power and root operations.
*   **Metric Properties:** For $p \ge 1$, Minkowski distance satisfies the triangle inequality, making it a valid metric for most hierarchical clustering linkage algorithms.

## **Jaccard distance:**

Jaccard distance is a metric used to measure the dissimilarity between two sets or binary vectors. It is the complement of the **Jaccard similarity coefficient** (also known as the Jaccard index).

### Mathematical Definition
If $A$ and $B$ are two sets, the Jaccard similarity is:
$$J(A, B) = \frac{|A \cap B|}{|A \cup B|}$$

The **Jaccard distance** is defined as:
$$d_J(A, B) = 1 - J(A, B) = 1 - \frac{|A \cap B|}{|A \cup B|}$$

### Key Characteristics
*   **Range:** The distance value always falls between **0 and 1**.
    *   **0:** The sets are identical ($A = B$).
    *   **1:** The sets have no elements in common (disjoint sets).
*   **Binary Data Application:** It is the standard metric for categorical or binary data (e.g., presence/absence of features, transaction histories, or DNA sequences).
*   **Non-Euclidean Nature:** Unlike Euclidean distance, Jaccard distance does not rely on the magnitude of values, but rather on the overlap of attributes.

### Use in Hierarchical Clustering
*   **Categorical Data:** It is preferred over Euclidean distance when working with sparse, high-dimensional binary datasets (like customer purchase patterns or text word counts) where "zeros" (the absence of a feature) should not be treated as a similarity.
*   **Linkage Compatibility:** It works effectively with most linkage methods (Single, Complete, Average) to group items based on the proportion of shared characteristics.

## **Hamming distance:**

Hamming distance is a metric used to measure the dissimilarity between two strings or vectors of equal length by counting the number of positions at which the corresponding symbols are different.

### Mathematical Definition
For two vectors $x$ and $y$ of length $n$:
$$d_H(x, y) = \sum_{i=1}^{n} [x_i \neq y_i]$$
Where $[x_i \neq y_i]$ is an indicator function that equals $1$ if the elements are different and $0$ if they are the same.

### Key Characteristics
*   **Requirement:** The two sequences must have the **same length**.
*   **Range:** The distance is an integer ranging from **$0$ to $n$** (where $n$ is the length of the vectors).
    *   **0:** The sequences are identical.
    *   **$n$:** The sequences are completely different (no matching elements at any position).
*   **Data Types:** Primarily used for **categorical data**, **binary strings** (bitstrings), or **textual data**.

### Common Applications
*   **Error Detection/Correction:** Used in telecommunications to identify the number of bit flips that occurred during transmission.
*   **Bioinformatics:** Comparing DNA or protein sequences to find mutations or variations.
*   **Natural Language Processing (NLP):** Comparing words or strings to identify typographical errors or spelling variations.

### Use in Hierarchical Clustering
*   **Categorical Clustering:** It is the preferred metric when the features are discrete/categorical rather than continuous.
*   **Sparse Data:** It is highly effective for high-dimensional binary data where the goal is to find items with the same "pattern" of attributes rather than the magnitude of those attributes.
*   **Linkage Sensitivity:** When used with hierarchical clustering, Hamming distance treats every mismatch as a discrete unit of distance, making the resulting dendrogram reflect the count of differing features.

## **Mahalanobis Distance:**

Mahalanobis distance is a measure of the distance between a point and a distribution, or between two points, that accounts for the **correlation** between variables and the **scale** of the data.

Unlike Euclidean distance, which treats all dimensions as independent and equally weighted, Mahalanobis distance scales the distance based on the covariance of the dataset.

### Mathematical Definition
The Mahalanobis distance $D_M$ between a vector $x$ and a vector $\mu$ is:
$$D_M(x, \mu) = \sqrt{(x - \mu)^T S^{-1} (x - \mu)}$$

Where:
*   **$x$**: The vector of the observation.
*   **$\mu$**: The vector of the mean (centroid) of the distribution.
*   **$S^{-1}$**: The inverse of the **covariance matrix** of the dataset.

### Key Characteristics
*   **Scale Invariance:** It is invariant to the scaling of the variables. If you multiply a feature by 10, the Mahalanobis distance remains unchanged because the covariance matrix adjusts accordingly.
*   **Correlation Awareness:** It accounts for how variables move together. If two variables are highly correlated, Mahalanobis distance "de-correlates" them, preventing the redundant information from over-weighting the distance calculation.
*   **Ellipsoidal Clusters:** While Euclidean distance assumes clusters are spherical, Mahalanobis distance assumes clusters are **ellipsoidal**, allowing it to capture the true shape of the data distribution.

### Comparison: Euclidean vs. Mahalanobis
| Feature | Euclidean Distance | Mahalanobis Distance |
| :--- | :--- | :--- |
| **Shape Assumption** | Spherical | Ellipsoidal |
| **Variable Correlation** | Ignores correlation | Accounts for correlation |
| **Scale Sensitivity** | Highly sensitive to scale | Scale invariant |
| **Computational Cost** | Low | High (requires matrix inversion) |

### Use in Hierarchical Clustering
*   **Multivariate Data:** It is highly effective when features are highly correlated (e.g., height and weight) or have vastly different units (e.g., age in years and income in dollars).
*   **Outlier Detection:** Because it measures how many standard deviations a point is from the mean, it is a standard tool for identifying multivariate outliers.
*   **Complexity Note:** Using Mahalanobis distance in hierarchical clustering is computationally intensive because the covariance matrix must be calculated and inverted, and the distance must be recalculated or adjusted as clusters merge.
