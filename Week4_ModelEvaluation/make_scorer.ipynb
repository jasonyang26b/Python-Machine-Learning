## **make_scorer:**

`make_scorer` is a utility function in `scikit-learn` used to convert a scoring function (which typically takes `y_true` and `y_pred`) into a callable object that can be used by scikit-learn's model selection tools, such as `cross_val_score` or `GridSearchCV`.

### Purpose
Standard scikit-learn evaluation metrics often require two arguments: the ground truth labels and the predicted labels. However, scikit-learn's optimization tools expect a scorer object that follows a specific signature: `scorer(estimator, X, y)`. `make_scorer` acts as a wrapper to bridge this gap.

### Syntax
```python
from sklearn.metrics import make_scorer

new_scorer = make_scorer(score_func, greater_is_better=True, **kwargs)
```

### Parameters
*   **`score_func`**: The metric function to be converted (e.g., `accuracy_score`, `f1_score`, `mean_squared_error`).
*   **`greater_is_better`**: 
    *   `True` (default): Used for metrics where higher values are better (e.g., Accuracy, F1, R²).
    *   `False`: Used for error/loss metrics where lower values are better (e.g., MSE, MAE).
*   **`**kwargs`**: Additional keyword arguments to be passed to the underlying `score_func`.

### Example Usage

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, make_scorer

# 1. Define a custom scorer for F1-score
# We specify greater_is_better=True because a higher F1 is better
f1_scorer = make_scorer(f1_score)

# 2. Use it in cross-validation
clf = RandomForestClassifier()
scores = cross_val_score(clf, X, y, scoring=f1_scorer)

print(scores.mean())
```

### Key Differences: Metric vs. Scorer
| Feature | Metric Function (e.g., `accuracy_score`) | Scorer Object (from `make_scorer`) |
| :--- | :--- | :--- |
| **Arguments** | `(y_true, y_pred)` | `(estimator, X, y)` |
| **Usage** | Manual evaluation of predictions | Passed to `GridSearchCV`, `cross_val_score` |
| **Logic** | Calculates error/accuracy | Calls `estimator.predict(X)` internally, then applies metric |

## Example:

````python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import make_scorer, accuracy_score
from sklearn.datasets import load_iris

# load data
data = load_iris()
X, y = data.data, data.target

# Initialize model
clf = RandomForestClassifier(n_estimators=100)

# make_score
accuracy = make_scorer(accuracy_score, greater_is_better=True)

# Perform 5-fold cross_validation using accuracy
scores = cross_val_score(clf, X, y, cv=5, scoring=accuracy)

print(f'Scores per fold: {scores}')
print(f'Mean Accuracy: {scores.mean():.2f}')
print(f'Standard Deviation: {scores.std():.2f}')
````
