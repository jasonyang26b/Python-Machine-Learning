### Feature scaling:

Feature scaling ensures that all input features contribute equally to the model's learning process by bringing them into a comparable range. Without scaling, features with larger numerical magnitudes will disproportionately influence the model's weight updates and loss calculations.

Key reasons for optimal performance include:

* **Gradient Descent Convergence:** In models like Linear Regression, Logistic Regression, and Neural Networks, scaling creates a more spherical (rather than elongated) cost function surface. This allows gradient descent to take a more direct path to the global minimum, preventing oscillations and significantly speeding up convergence.
* **Distance-Based Algorithms:** Algorithms that rely on Euclidean distance or other proximity metrics—such as K-Nearest Neighbors (KNN), K-Means Clustering, and Support Vector Machines (SVM)—are highly sensitive to scale. If one feature ranges from 0–1 and another from 0–1,000, the larger feature will dominate the distance calculation, rendering the smaller feature irrelevant.
* **Regularization Fairness:** Regularization techniques (L1/Lasso and L2/Ridge) penalize the magnitude of coefficients. If features are on different scales, the penalty will be applied unfairly; features with smaller scales will require larger coefficients to have an effect, making them more susceptible to being penalized or zeroed out by the regularization term.
* **Numerical Stability:** Scaling prevents issues with floating-point precision and arithmetic overflow/underflow that can occur when performing matrix multiplications on extremely large or small values.
