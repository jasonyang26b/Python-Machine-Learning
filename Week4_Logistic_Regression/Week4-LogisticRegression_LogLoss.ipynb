## **Where is the logloss from?**

Log loss (also known as **Binary Cross-Entropy Loss**) is derived from the principle of **Maximum Likelihood Estimation (MLE)** applied to a Bernoulli distribution.

Here is the step-by-step derivation:

### 1. The Bernoulli Distribution
In binary classification, we assume each outcome $y$ (where $y \in \{0, 1\}$) follows a Bernoulli distribution. The probability mass function (PMF) for a single observation is:
$$P(y|p) = p^y (1-p)^{1-y}$$

Where:
*   $p$ is the probability predicted by the model (e.g., via a Sigmoid function).
*   If $y=1$, the formula becomes $p^1(1-p)^0 = p$.
*   If $y=1$, the formula becomes $p^0(1-p)^1 = 1-p$.

### 2. The Likelihood Function
For a dataset of $n$ independent observations, the **Likelihood** $L(p)$ is the product of the individual probabilities:
$$L(p) = \prod_{i=1}^{n} p_i^{y_i} (1-p_i)^{1-y_i}$$

The goal of the model is to find the parameters that maximize this likelihood.

### 3. The Log-Likelihood
Maximizing a product is mathematically difficult and computationally unstable (multiplying many small probabilities leads to arithmetic underflow). To simplify, we take the **natural logarithm** of the likelihood. Because the logarithm is a monotonic function, the value that maximizes the likelihood also maximizes the log-likelihood.

$$\log(L(p)) = \log \left( \prod_{i=1}^{n} p_i^{y_i} (1-p_i)^{1-y_i} \right)$$

Using the property $\log(a \cdot b) = \log(a) + \log(b)$ and $\log(a^b) = b \log(a)$, we get:
$$\log(L(p)) = \sum_{i=1}^{n} [y_i \log(p_i) + (1-y_i) \log(1-p_i)]$$

### 4. The Negative Log-Likelihood (Loss Function)
In optimization, we typically perform **Gradient Descent**, which seeks to *minimize* a function rather than maximize it. To turn the maximization problem into a minimization problem, we multiply the log-likelihood by $-1$. 

This gives us the **Negative Log-Likelihood (NLL)**, which is exactly the formula for **Log Loss**:

$$\text{Log Loss} = -\sum_{i=1}^{n} [y_i \log(p_i) + (1-p_i) \log(1-p_i)]$$

*(Note: Often, we take the average over $n$ samples, dividing by $n$, to make the loss independent of the dataset size.)*

### Summary of Logic
1.  **Model the data:** Assume outcomes follow a Bernoulli distribution.
2.  **Likelihood:** Write the probability of seeing the actual labels given the predicted probabilities.
3.  **Logarithm:** Convert the product into a sum to make it mathematically manageable.
4.  **Negation:** Flip the sign so we can use minimization algorithms (Gradient Descent) to find the best weights.
