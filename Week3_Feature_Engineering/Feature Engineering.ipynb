## **Feature Engineering:**

Feature engineering is the process of using domain knowledge to select, manipulate, and transform raw data into meaningful features that improve the performance of machine learning algorithms. It involves converting data into a format that makes it easier for models to identify underlying patterns.

Key techniques include:

*   **Imputation:** Handling missing values by replacing them with the mean, median, mode, or a constant.
*   **Encoding:** Converting categorical data into numerical formats (e.g., One-Hot Encoding or Label Encoding).
*   **Scaling:** Normalizing or standardizing numerical values to a specific range (e.g., 0 to 1) to ensure features contribute equally to the model.
*   **Transformation:** Applying mathematical functions like log transforms to reduce skewness or stabilize variance.
*   **Feature Extraction:** Creating new features from existing ones (e.g., extracting "day of the week" from a timestamp or "BMI" from height and weight).
*   **Feature Selection:** Identifying and retaining only the most relevant features to reduce dimensionality and prevent overfitting.
*   **Handling Outliers:** Detecting and managing extreme values that could skew model training.

## **What is statsmodels.api:**

`statsmodels.api` is a Python library used for estimating and testing statistical models, such as linear regression, generalized linear models, and time-series analysis. Unlike Scikit-learn, which focuses on predictive accuracy, `statsmodels` focuses on statistical inference and detailed model summaries.

### 2. Linear Regression (OLS)
The most common use case is Ordinary Least Squares (OLS) regression. Note that `statsmodels` does **not** automatically include an intercept (constant) term; you must add it manually.

```python
import statsmodels.api as sm
import numpy as np

# Sample data
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 11])

# Add a constant column for the intercept
X = sm.add_constant(X)

# Fit the model
model = sm.OLS(y, X).fit()

# View the detailed statistical summary
print(model.summary())
```

### 3. Interpreting the Summary
When you call `model.summary()`, the output provides critical statistical metrics:
*   **R-squared:** How much of the variance in $y$ is explained by $X$.
*   **P-values ($P > |t|$):** If a p-value is $< 0.05$, the feature is considered statistically significant.
*   **Coefficients (coef):** The estimated weight of each feature.
*   **Standard Error:** The precision of the coefficient estimate.

### 4. Generalized Linear Models (GLM)
Used for data that doesn't follow a normal distribution (e.g., binary outcomes using Logistic Regression).

```python
# Logistic Regression example
y_binary = np.array([0, 0, 0, 1, 1])
X_binary = sm.add_constant(np.array([1, 2, 3, 4, 5]))

logit_model = sm.GLM(y_binary, X_binary, family=sm.families.Binomial()).fit()
print(logit_model.summary())
```

### 5. Time-Series Analysis (TSA)
`statsmodels` is the industry standard for time-series forecasting using models like ARIMA or SARIMA.

```python
from statsmodels.tsa.arima.model import ARIMA

# Sample time-series data
data = [10, 12, 15, 14, 18, 20, 22, 21]

# Fit an ARIMA model (p, d, q)
model = ARIMA(data, order=(1, 1, 1))
model_fit = model.fit()

# Forecast the next 3 steps
forecast = model_fit.forecast(steps=3)
print(forecast)
```

| Feature | `statsmodels` | `scikit-learn` |
| :--- | :--- | :--- |

| **Complexity** | High statistical rigor | High predictive power |
| **Best for** | Hypothesis testing, scientific research | Machine learning pipelines, production |
